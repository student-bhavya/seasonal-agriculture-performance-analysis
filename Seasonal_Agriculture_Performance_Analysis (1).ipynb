## 9. Outlier Analysis

```python
outlier_summary = []

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    outlier_summary.append({
        'Column': col,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': len(outliers),
        'Outlier Percentage': round((len(outliers) / len(df)) * 100, 2)
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary = outlier_summary.sort_values(
    by='Outlier Count',
    ascending=False
)

outlier_summary
```

## 14. Key Insights

### Insight 1: Kharif Shows the Strongest Overall Performance

**Observation:** Kharif recorded the highest average yield, profit, and water efficiency among the three seasons.

**Evidence:** Kharif had an average yield of 5.63 tonnes/ha, average profit of ₹178,915, and water efficiency of 5.89 t per 1000 m³. Rabi recorded 5.04 tonnes/ha yield and ₹87,689 profit, while Zaid recorded 4.64 tonnes/ha yield and an average loss of ₹24,805.

**Interpretation:** The results indicate that Kharif performed better across the selected agricultural and economic performance measures in this dataset.

**Limitation:** These results show seasonal association in the dataset and do not prove that the season alone caused the differences.

### Insight 2: Kharif Yield Is Higher Across All Crops

**Observation:** Every crop in the dataset recorded its highest average yield during Kharif and its lowest average yield during Zaid.

**Evidence:** For example, Sugarcane averaged 53.46 tonnes/ha in Kharif, 43.29 in Rabi, and 38.42 in Zaid. Maize averaged 2.97, 2.60, and 2.29 tonnes/ha respectively.

**Interpretation:** The seasonal pattern is consistent across all eight crops, with Kharif showing better average yield than Rabi and Zaid.

**Limitation:** This pattern is based on the observed dataset and does not establish that season alone determines crop yield.

### Insight 3: Agricultural Performance Varies Across States

**Observation:** Average agricultural performance differs across the eight states in the dataset.

**Evidence:** Punjab recorded the highest average yield at 6.12 tonnes/ha, while Andhra Pradesh recorded the lowest at 4.63 tonnes/ha. Punjab also had the highest average water efficiency at 5.62 t per 1000 m³.

**Interpretation:** The results indicate noticeable regional variation in agricultural performance and resource efficiency.

**Limitation:** The analysis does not identify the specific reasons for these state-level differences, and the results should not be interpreted as showing that location alone causes higher or lower performance.

### Insight 4: Flood Irrigation Has the Highest Average Water Usage

**Observation:** Average water usage differs considerably across irrigation methods.

**Evidence:** Flood irrigation recorded the highest average water usage at 8,026 m³, while Rainfed farming recorded the lowest at 3,550 m³. Sprinkler and Drip recorded 6,208 m³ and 5,919 m³ respectively.

**Interpretation:** The results show that irrigation practices are associated with different levels of water consumption in the dataset, with Flood irrigation having the highest average usage.

**Limitation:** Water usage can also depend on factors such as farm conditions, crop type, and season, so irrigation method alone cannot be considered the sole reason for the difference.

### Insight 5: Yield Shows a Strong Positive Correlation with Water Efficiency

**Observation:** Yield and water efficiency show a strong positive correlation in the dataset.

**Evidence:** The correlation coefficient between Yield_Tonnes_Ha and Water_Efficiency_t_per_1000m3 is approximately 0.91.

**Interpretation:** Farms with higher recorded yield generally also show higher water efficiency in this dataset.

**Limitation:** Correlation indicates an association between the variables and does not prove that higher yield directly causes higher water efficiency.

### Insight 6: Revenue Shows a Strong Positive Relationship with Profit

**Observation:** Revenue and profit have a strong positive correlation in the dataset.

**Evidence:** The correlation coefficient between Revenue_INR and Profit_INR is approximately 0.89.

**Interpretation:** Farms with higher recorded revenue generally tend to have higher recorded profit.

**Limitation:** Profit also depends on total costs and other economic factors, so the correlation does not mean that higher revenue alone causes higher profit.

### Insight 7: Yield Has a Strong Positive Correlation with Production

**Observation:** Yield and production show a strong positive correlation in the dataset.

**Evidence:** The correlation coefficient between Yield_Tonnes_Ha and Production_Tonnes is approximately 0.88.

**Interpretation:** Farms with higher yield per hectare generally tend to have higher total production in the dataset.

**Limitation:** Total production can also depend on farm area, so yield alone does not determine production.

### Insight 8: Rainfall Shows a Positive Correlation with Disease/Pest Risk

**Observation:** Rainfall and disease/pest risk show a positive correlation in the dataset.

**Evidence:** The correlation coefficient between Rainfall_mm and Disease_Pest_Risk_pct is approximately 0.62.

**Interpretation:** Higher recorded rainfall levels are generally associated with higher recorded disease/pest risk in this dataset.

**Limitation:** The correlation does not prove that rainfall directly causes higher disease or pest risk, as other environmental and farming factors may also contribute.

## 16. Limitations

2. The analysis mainly identifies patterns and associations; it does not establish causal relationships between variables.

3. Some variables contain unusual or extreme observations, particularly yield, production, revenue, profit, and water efficiency. These observations were investigated using the IQR method but were retained because they were not confirmed as data errors.

4. The analysis uses descriptive statistics and visualizations and does not include predictive machine learning models.
